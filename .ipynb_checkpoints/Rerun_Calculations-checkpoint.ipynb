# Overview

This file is provided as a demo of the full algorithm.

It reruns all calculations from raw data to final figure:
- *Utilities/Data_Preparation_ERA5_025_Restricted_all.py* processes observed and simulated (GCM) data.
- *Utilities/Full_run_Model.py* applies the algorithm from RR2020 (Simulated prior and 2-steps constraint with observations)
- *Utilities/Figures.py* produces figures and indexes, including the predictive sampling


This produces the same figures as *Projections_TXxx.ipynb* (Figures 2 and Tables 1, 2 and Supplementary Figures 7, 8 , 9, 10 and 11 and Tables 4, 5) and adds Supplementary Figure 6 and Table 3 (from data preparation).

Expect a total run time of up to 10 minutes (tested on a laptop)

```python
import sys,os
sys.version
```

```python
## Input data (raw)

### ERA5 0.25.25
### TXX World

ref_deb=1986
ref_fin=2016

Y_ref="World"
X_ref="World"
station_type=False

work_path=os.path.abspath(os.getcwd())+"/Rerun_World/Era5_025"
path_data=os.path.abspath(os.getcwd())+"/Data_origin/"

path_X_Modeles_Brut=path_data+"/CMIP6_TasMean_"+X_ref
path_Y_Modeles_Brut=path_data+"/CMIP6_TXX_"+Y_ref

path_X_Obs_Brut=path_data+"/ERA5_Obs_TasMean_"+X_ref+"/ERA5_tas_"+X_ref+".nc"
path_Y_Obs_Brut=path_data+"/ERA5_025_Obs_TXX_World/ERA5_tasmax_World.nc"

pathObs=work_path+"/data/Observations/03_Post_Treatment/"+Y_ref+"_"+X_ref
pathGCM_X_separated=work_path+"/data/CMIP6/03_Post_Treatment/TasMean_"+X_ref+"/separated"
pathGCM_Y_separated=work_path+"/data/CMIP6/03_Post_Treatment/TXX_"+Y_ref+"/separated"


```

```python
%%time

### Run once - All model calculations
!python3 Utilities/Data_Preparation_ERA5_025_Restricted_all.py $Y_ref $X_ref $work_path $path_X_Modeles_Brut $path_Y_Modeles_Brut $path_X_Obs_Brut $path_Y_Obs_Brut $station_type
results =!python3 Utilities/Full_run_Model.py $Y_ref $X_ref $work_path $pathObs $pathGCM_X_separated $pathGCM_Y_separated $ref_deb $ref_fin
dt_string=results[-1]
print(results)
!python3 Utilities/Figures.py $Y_ref $X_ref $work_path $pathObs $pathGCM_X_separated $pathGCM_Y_separated $dt_string
```
